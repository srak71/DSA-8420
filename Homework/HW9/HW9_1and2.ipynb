```python
import numpy as np
from scipy.optimize import linprog
```

![Dual Simplex Image](P1_DS.png)

```python
c = -1 * np.array([-8, -2, -12]) 
A = np.array([[-2, -1, -4], 
              [-1, 2, -2], 
              [0, -1, 1]])
b = np.array([1, -6, 2])
x_bounds = np.array([(0, None), (0, None), (0, None)])

# dual simplex
solution = linprog(c, A_ub= A, b_ub= b, bounds= x_bounds, method= 'highs-ds')

print(solution.message)
print(f'x1= {solution.x[0]}')
print(f'x2= {solution.x[1]}')
print(f'x3= {solution.x[2]}')
print(f'max Z: {-solution.fun}') 
```

![Dual Simplex 2](P2_DS.png)

```python
c = -1 * np.array([-8, -1, -1]) 
A = np.array([[-2, -1, 0], 
              [2, -1, 1], 
              [1, 0, 1], 
              [-1, 1, -1]])
b = np.array([-20, -12, 7, -5])
x_bounds = np.array([(0, None), (0, None), (0, None)])

sol = linprog(c, A_ub= A, b_ub= b, bounds= x_bounds, method= "highs-ds")

print(sol.message)
```
