# 1C.

```python
from scipy.optimize import linprog

c = [-2, 4, -7]
A_ub = [[1, 1, 2], [-1, -1, 1]] # x1 + x2 + 2x3 ≤ 10, -x1 - x2 + x3 ≤ -6
b_ub = [10, -6]

A_eq = [[1, -1, 1]] # x1 - x2 + x3 = 2
b_eq = [2]

#  x1 >= 0, x2 >= 0, x3 >= 0
x_bounds = [(0, None), (0, None), (0, None)]

result = linprog(c, A_ub=A_ub, b_ub=b_ub, A_eq=A_eq, b_eq=b_eq, bounds=x_bounds, method='simplex')

print(f"Optimal x1: {result.x[0]}")
print(f"Optimal x2: {result.x[1]}")
print(f"Optimal x3: {result.x[2]}")
print(f"Optimal value: {-result.fun}")
```

```python
c_phase2 = [-2, -1, 0, 0, 0]
# x1 - x2 - s1 = 8, 2x1 + 3x2 + s2 = 24, 2x1 + x2 + s3 = 12
A_eq_phase2 = [[1, -1, -1, 0, 0], [2, 3, 0, 1, 0], [2, 1, 0, 0, 1]]
b_eq_phase2 = [8, 24, 12]

x_bounds_phase2 = [(0, None), (0, None), (0, None), (0, None), (0, None)]

result_phase2 = linprog(c_phase2, A_eq=A_eq_phase2, b_eq=b_eq_phase2, bounds=x_bounds_phase2, method="simplex")

if not result_phase2.success:
    print("Problem is infeasable!!!")
else:
    print(f"Optimal x1: {result_phase2.x[0]}")
    print(f"Optimal x2: {result_phase2.x[1]}")
    print(f"Optimal x3: {result_phase2.x[2]}")
    print(f"Optimal value: {result_phase2.fun}")
```
