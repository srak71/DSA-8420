As we know, an assignment problem is a speical linear program. In this project, we solve an assignment problem using the scipy.optimize package in two ways -- as a general linear program and as an assignment problem. 

```python
import numpy as np
from scipy.optimize import linprog, linear_sum_assignment
import time
```

```python
n = 200   # n: number of jobs/workers
cost = np.random.randint(0, n**2, (n, n))   # generate an n-by-n random matrix with integer entries from 0 to n^2-1
```

# Task 1: (35 Points)

Let C = [$c_{ij}$] be the cost matrix of an assignment problem. The problem can be modeled as a linear program:

$min\Sigma_{i=1}^{n} \Sigma_{j=1}^{n} (c_{ij} x_{ij})$

subject to:

$\Sigma_{i=1}^{n} x_{ij} = 1$ for $j = 1, ... , n$

$\Sigma_{j=1}^{n} x_{ij} = 1$ for $i = 1, ... , n$

$x_{ij} \geq 0$ for $i, j = 1 , ... , n$

```python
c = cost.reshape((n**2,1))   # reshape the cost matrix to a column vector
```

Define A_eq and b_eq in the following cell. You may use multiple cells if needed. Your code should work for a general n. Use the following documents and/or other online resources for help.

- **scipy.optimize.linprog**: https://docs.scipy.org/doc/scipy/reference/generated/scipy.optimize.linprog.html

- **Numpy basics**: https://numpy.org/doc/stable/user/absolute_beginners.html

Constraints:

- Each worker does exactly one job: $\sum_{j=1}^{n} x_{ij} = 1$ for each worker $i$ (total of n constraints).

- Each job is assigned to exactly one worker: $\sum_{i=1}^{n} x_{ij} = 1$ for each job $j$ (total of n constraints).

We have 2n constraints in total, each constraint is a row in A_eq... should have dimensions (2n, n**2).

- Worker constraint: sum over each row = 1.

- Job constraint: sum over each col = 1.

```python
# A_eq
A_eq = np.zeros((2*n, n**2))

# worker constraint
for i in range(n):
    for j in range(n):
        A_eq[i, n*i+j] = 1
        
# job constraint
for j in range(n):
    for i in range(n):
        A_eq[n+j, n*i+j] = 1
        
# RHS
b_eq = np.ones(2*n)
```

```python
start_time = time.time()   # start timing

res = linprog(c, None, None, A_eq, b_eq)

end_time = time.time()   # end timing
elapsed_LP = end_time - start_time
```

```python
print('----  linprog()  ----')
print(f"Minimum cost: {res.fun}.")
print(f"Completed in {elapsed_LP} s.")
```

# Task 2: (15 points)

Instead of solving the assignment problem as a linear program, the function *scipy.optimize.linear_sum_assignment()* solves the assignment problem directly. Your code should solve the problem with the cost matrix cost and record the elapsed time. If coded correctly, the minimum costs returned by the two methods should be identical (up to some tiny numerical error).

Use the following cell to solve the assignment problem with the cost matrix "cost" and record the elapsed time. You may use multiple cells if needed.

```python
start_time = time.time()
row_ind, col_ind = linear_sum_assignment(cost) # comput total cost
cur_time = time.time() - start_time
min_cost = cost[row_ind, col_ind].sum() # compute min cost

print('----  linear_sum_assignment()  ----')
print(f'Minimum cost = {min_cost}.')
print(f'Completed in {cur_time} s.')
```
